```python
import pandas as pd
```

```python
df = pd.read_csv('insurance.csv')  #Load the Data
```

```python
df.shape, df.head()
```

```python
df.describe(include='all').T # Descriptive Statistics
```

```python
df.info()
df.isnull().sum().sort_values(ascending=False)   # Data Types and Null Check
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print(df.columns.tolist())
```

```python
numerical = ['age', 'bmi', 'children', 'charges']
for col in numerical:
    plt.figure(figsize=(6, 4))
    sns.histplot(df[col].dropna(), kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show() # Univariate Analysis:Plotting Distributions for Key Numeric Columns
```

```python
categorical = ['sex', 'smoker', 'region']
for col in categorical:
    plt.figure(figsize=(6, 4))
    sns.countplot(x=col, data=df)
    plt.title(f'Count of {col}')
    plt.xticks(rotation=45)
    plt.show()
 #Count Plots for Categorical Features
```

```python
#Bivariate Analysis: Loss Ratio by Group 
df.groupby('region')['charges'].mean().plot(kind='bar')
plt.title('Average Charges by Region')
plt.ylabel('Average Charges')
plt.show()
```

```python
sns.boxplot(x='smoker', y='charges', data=df)
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df[numerical].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Between Key Financial Variables")
plt.show()
#Correlation Matrix
```

```python
for col in numerical:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
#Outlier Detection
```

```python
 import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x='smoker', y='charges', data=df)
plt.title('Distribution of Charges by Smoker Status')
plt.show()
#Creative Insights Plots: Average Charges by Smoker Status
```

```python
df.groupby('region')['charges'].mean().plot(kind='bar', color='skyblue')
plt.title('Average Charges by Region')
plt.ylabel('Average Charges')
plt.xticks(rotation=45)
plt.show()
#Average Charges by Region
```

```python
sns.barplot(x='children', y='charges', data=df)
plt.title('Average Charges by Number of Children')
plt.ylabel('Average Charges')
plt.show()
#Charges by Number of Children
```

```python
sns.scatterplot(x='bmi', y='charges', hue='smoker', data=df)
plt.title('Charges vs BMI (Colored by Smoker)')
plt.show()
#Charges vs BMI Scatterplot
```
