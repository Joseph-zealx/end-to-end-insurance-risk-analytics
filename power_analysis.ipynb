```python

# Parameters for power analysis
effect_size = 0.5  # Medium effect size (Cohen's d)
alpha = 0.05       # Significance level
sample_size = 64   # Sample size per group

# Calculate the power of the test
power = sm.stats.tt_ind_solve_power(effect_size=effect_size, 
                                     alpha=alpha, 
                                     nobs1=sample_size, 
                                     alternative='two-sided')

print(f"Calculated power of the test: {power:.4f}")
```
